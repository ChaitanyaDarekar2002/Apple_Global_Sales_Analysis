```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('apple_global_sales_dataset.csv')
data
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
(data.isnull().sum()/len(data))*100
```

```python
data["customer_rating"] = data["customer_rating"].fillna(data["customer_rating"].median())
```

```python
data.duplicated().sum()
```

```python
data.isnull().sum()
```

```python
data["previous_device_os"] = data["previous_device_os"].fillna(
    data["previous_device_os"].mode()[0]
)
```

```python
data.isnull().sum()
```

```python
data["storage"] = data["storage"].fillna(
    data["storage"].mode()[0]
)
```

```python
data.isnull().sum()
```

```python
sns.boxplot(x=data["revenue_usd"])
plt.show()
```

```python
numeric = data.select_dtypes(include="number")

plt.figure(figsize=(10,7))
sns.heatmap(numeric.corr(),
            annot=True,
            cmap="coolwarm")
plt.show()
```

```python
sns.histplot(data["revenue_usd"], bins=30)
plt.show()
```

```python
top_products = (data.groupby("product_name")["units_sold"].sum().sort_values(ascending=False).head(10))

top_products.plot(kind="bar")
plt.title("Top 10 Products")
plt.show()
```

```python
category = data.groupby("category")["revenue_usd"].sum()

category.plot(kind="bar",color='orange')
plt.show()
```

```python
country = (data.groupby("country")["revenue_usd"].sum().sort_values(ascending=False).head(10))

country.plot(kind="bar")
plt.show()
```

```python
country = (data.groupby("country")["unit_price_usd"].sum().sort_values(ascending=False).head(10))

country.plot(kind="line")
plt.show()
```

```python
channel = data.groupby("sales_channel")["revenue_usd"].sum()

channel.plot(kind="pie", autopct='%1.1f%%')
```

```python
payment = data.groupby("payment_method")["revenue_usd"].sum()

payment.plot(kind="bar",color='orange')
```

```python
segment = data.groupby("customer_segment")["revenue_usd"].sum()

segment.plot(kind="line")
plt.show()
```

```python
sns.countplot(data=data,x="customer_rating")
plt.xticks(rotation=45)
plt.show()
```

```python
returns = data["return_status"].value_counts()

returns.plot(kind="pie",
             autopct='%1.1f%%')
plt.ylabel("")
plt.show()
```

```python
sns.scatterplot(
    data=data,
    x="units_sold",
    y="revenue_usd"
)

plt.show()
```

```python
sns.pairplot(
    data,
    vars=["unit_price_usd", "discount_pct", "units_sold", "revenue_usd"],
    hue="category",
    palette="coolwarm"
)
plt.show()
```

```python
data.to_csv("Apple_Sales_Clean_Data.csv", index=False)
```
